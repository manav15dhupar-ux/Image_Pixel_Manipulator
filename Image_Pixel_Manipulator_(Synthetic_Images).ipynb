```python
import numpy as np
import matplotlib.pyplot as plt
```


# Canvas

**Box**

```python
img = np.zeros((200,200,3), dtype=np.uint8)

img[:, :] = [200, 230, 255]

img[10:50, 10:50] = [255, 0, 0]
img[60:100, 10:50] = [0, 255, 0]
img[110:150, 10:50] = [0, 0, 255]
img[10:50, 70:110] = [255, 255, 0]
img[60:100, 70:110] = [255, 0, 255]
img[110:150, 70:110] = [0, 255, 255]
img[10:50, 130:170] = [0, 0, 0]
img[60:100, 130:170] = [255, 255, 255]

plt.imshow(img)
plt.title("Colorfull Box")
plt.axis("off")
plt.show()
```

**Stripes**

```python
img = np.zeros((200,200,3), dtype=np.uint8)

img[:, :] = [200, 230, 255]

img[10:40, :] = [255, 0, 0]
img[50:80, :] = [0, 255, 0]
img[90:120, :] = [0, 0, 255]
img[130:160, :] = [0, 0, 0]
img[170:200, :] = [255, 255, 0]

plt.imshow(img)
plt.title("Colourful Stripes")
plt.axis("off")
plt.show()
```

**Crop**

```python
img = np.zeros((200, 200, 3), dtype=np.uint8)

img[:, :] = [200, 230, 255]

img[10:50, 10:50] = [255, 0, 0]
img[60:100, 10:50] = [0, 255, 0]
img[110:150, 10:50] = [0, 0, 255]

crop = img[20:40, 20:40]

print(f"Orginal Shape: {img.shape}")
print(f"Croped Shape: {crop.shape}")
print()

plt.imshow(img)
plt.title("Orginal")
plt.axis("off")
plt.show()

plt.imshow(crop)
plt.title("Croped")
plt.axis("off")
plt.show()
```

**Flip**

```python
flipped_horizontal = img[:, ::-1]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(flipped_horizontal)
plt.title("Horizontal Flip")
plt.axis("off")

plt.show()
```

```python
flipped_vertical = img[::-1, :]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(flipped_vertical)
plt.title("Vertical Flip")
plt.axis("off")

plt.show()
```
