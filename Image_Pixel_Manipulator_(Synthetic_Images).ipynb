```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((200,200,3), dtype=np.uint8)

img[50:100,50:100] = [255,0,0]
img[50:100,125:175] = [0,255,0]
img[125:175,50:100] = [0,0,255]
img[125:175,125:175] = [255,255,0]

plt.imshow(img)
plt.show()
```
